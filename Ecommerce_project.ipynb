```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
#
```

```python
customers_raw=pd.read_csv("/content/customers_raw.csv")
```

```python
orders_raw=pd.read_csv("/content/orders_raw.csv")
```

```python
order_items_raw=pd.read_csv("/content/order_items_raw.csv")
#
```

```python
# Task A — Understand the data: inspect shape, columns, data types, sample rows, and summary statistics

print("=== CUSTOMERS DATASET ===")
print(f"Shape: {customers_raw.shape}")
print("\nData Types:")
print(customers_raw.dtypes)
print("\nSample Rows:")
display(customers_raw.head())
print("\nSummary Statistics:")
display(customers_raw.describe(include='all'))



```

```python
print("=== ORDERS DATASET ===")
print(f"Shape: {orders_raw.shape}")
print("\nData Types:")
print(orders_raw.dtypes)
print("\nSample Rows:")
display(orders_raw.head())
print("\nSummary Statistics:")
display(orders_raw.describe(include='all'))




```

```python
print("=== ORDER ITEMS DATASET ===")
print(f"Shape: {order_items_raw.shape}")
print("\nData Types:")
print(order_items_raw.dtypes)
print("\nSample Rows:")
display(order_items_raw.head())
print("\nSummary Statistics:")
display(order_items_raw.describe(include='all'))
```

```python
# Task B — Clean the data separately for each table using exact column names

# 1. Cleaning CUSTOMERS
customers_clean = customers_raw.copy()
# Drop duplicate customer IDs to keep primary key unique
customers_clean = customers_clean.drop_duplicates(subset=['Customer_ID'])
# Impute missing values
customers_clean['City'] = customers_clean['City'].fillna('Unknown')
customers_clean['Customer_Segment'] = customers_clean['Customer_Segment'].fillna('Unknown')
# Ensure proper dates
customers_clean['Signup_Date'] = pd.to_datetime(customers_clean['Signup_Date'], errors='coerce')

# 2. Cleaning ORDERS
orders_clean = orders_raw.copy()
orders_clean = orders_clean.drop_duplicates(subset=['Order_ID'])
orders_clean['Order_Date'] = pd.to_datetime(orders_clean['Order_Date'], errors='coerce')
orders_clean['Delivery_Date'] = pd.to_datetime(orders_clean['Delivery_Date'], errors='coerce')
# Fill missing categorical features
orders_clean['Sales_Channel'] = orders_clean['Sales_Channel'].fillna('Unknown')
orders_clean['Payment_Method'] = orders_clean['Payment_Method'].fillna('Unknown')
orders_clean['Order_Status'] = orders_clean['Order_Status'].fillna('Unknown')
orders_clean['Shipping_Fee_EGP'] = orders_clean['Shipping_Fee_EGP'].fillna(0.0)

# CRITICAL: Exclude both 'Cancelled' and 'Returned' orders to analyze real Net Sales/Revenue
orders_clean = orders_clean[~orders_clean['Order_Status'].isin(['Cancelled', 'Returned'])]

# 3. Cleaning ORDER ITEMS
order_items_clean = order_items_raw.copy()
# Clean and format numeric fields
order_items_clean['Quantity'] = pd.to_numeric(order_items_clean['Quantity'], errors='coerce').fillna(1).astype(int)
order_items_clean['Unit_Price_EGP'] = pd.to_numeric(order_items_clean['Unit_Price_EGP'], errors='coerce').fillna(0.0)
order_items_clean['Discount_Rate'] = pd.to_numeric(order_items_clean['Discount_Rate'], errors='coerce').fillna(0.0)
order_items_clean['Unit_Cost_EGP'] = pd.to_numeric(order_items_clean['Unit_Cost_EGP'], errors='coerce').fillna(0.0)

# Recalculate financial fields to guarantee mathematical accuracy
order_items_clean['Line_Sales_EGP'] = (order_items_clean['Unit_Price_EGP'] * order_items_clean['Quantity']) * (1.0 - order_items_clean['Discount_Rate'])
order_items_clean['Line_Cost_EGP'] = order_items_clean['Unit_Cost_EGP'] * order_items_clean['Quantity']
order_items_clean['Line_Profit_EGP'] = order_items_clean['Line_Sales_EGP'] - order_items_clean['Line_Cost_EGP']

# 4. Referential Integrity
# Filter out order items referencing non-existent orders (and now also filtered-out Cancelled & Returned orders)
valid_orders = set(orders_clean['Order_ID'])
order_items_clean = order_items_clean[order_items_clean['Order_ID'].isin(valid_orders)]

# Print cleaning summaries
print(f"Customers clean shape: {customers_clean.shape} (Original: {customers_raw.shape})")
print(f"Orders clean shape: {orders_clean.shape} (Original (with Cancelled & Returned): {orders_raw.shape})")
print(f"Order items clean shape: {order_items_clean.shape} (Original: {order_items_raw.shape})")
print("Referential integrity successfully checked & verified (Cancelled & Returned orders successfully excluded).")
```

```python
# Task C — Explore and Analyze

# 1. Correct Joins
# Join Orders and Customers (Order-level data)
orders_customers = pd.merge(orders_clean, customers_clean, on='Customer_ID', how='inner')

# Calculate Order Total Sales & Profit from order items to safely join back to orders without double counting
order_metrics = order_items_clean.groupby('Order_ID').agg(
    Total_Order_Sales_EGP=('Line_Sales_EGP', 'sum'),
    Total_Order_Cost_EGP=('Line_Cost_EGP', 'sum'),
    Total_Order_Profit_EGP=('Line_Profit_EGP', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).reset_index()

orders_summary = pd.merge(orders_customers, order_metrics, on='Order_ID', how='inner')

# Time analysis preparation
orders_summary['Order_Month'] = orders_summary['Order_Date'].dt.to_period('M')
```

```python
# Export updated datasets excluding cancelled orders to CSVs for persistence
customers_clean.to_csv('customers_clean.csv', index=False)
orders_clean.to_csv('orders_clean.csv', index=False)
order_items_clean.to_csv('order_items_clean.csv', index=False)
orders_customers.to_csv('orders_customers.csv', index=False)
order_metrics.to_csv('order_metrics.csv', index=False)
orders_summary.to_csv('orders_summary.csv', index=False)
print("All clean and updated datasets exported to CSV successfully!")
```

```python


# 2. Answer Question 1: Order volume and sales by month, channel, city, and customer segment
print("=== Q1: Order Volume & Sales Analysis (Excluding Cancelled & Returned) ===")
month_analysis = orders_summary.groupby('Order_Month').agg(Order_Count=('Order_ID', 'count'), Total_Sales_EGP=('Total_Order_Sales_EGP', 'sum')).reset_index()
display(month_analysis)

channel_analysis = orders_summary.groupby('Sales_Channel').agg(Order_Count=('Order_ID', 'count'), Total_Sales_EGP=('Total_Order_Sales_EGP', 'sum')).reset_index()
display(channel_analysis)

city_analysis = orders_summary.groupby('City').agg(Order_Count=('Order_ID', 'count'), Total_Sales_EGP=('Total_Order_Sales_EGP', 'sum')).reset_index()
display(city_analysis)

segment_analysis = orders_summary.groupby('Customer_Segment').agg(Order_Count=('Order_ID', 'count'), Total_Sales_EGP=('Total_Order_Sales_EGP', 'sum')).reset_index()
display(segment_analysis)
```

```python
# 3. Answer Question 2: Which product categories and products contribute most to sales and profit?
print("\n=== Q2: Product & Category Analysis ===")
category_contrib = order_items_clean.groupby('Category').agg(
    Total_Sales_EGP=('Line_Sales_EGP', 'sum'),
    Total_Profit_EGP=('Line_Profit_EGP', 'sum')
).sort_values(by='Total_Profit_EGP', ascending=False).reset_index()
display(category_contrib)

product_contrib = order_items_clean.groupby('Product').agg(
    Total_Sales_EGP=('Line_Sales_EGP', 'sum'),
    Total_Profit_EGP=('Line_Profit_EGP', 'sum')
).sort_values(by='Total_Profit_EGP', ascending=False).head(10).reset_index()
display(product_contrib)

```

```python
# 4. Answer Question 3: Order status and delivery timing across segments / channels
orders_summary['Delivery_Time_Days'] = (orders_summary['Delivery_Date'] - orders_summary['Order_Date']).dt.days

print("\n=== Q3: Order Status & Delivery Timing ===")
delivery_by_segment = orders_summary.groupby('Customer_Segment').agg(
    Avg_Delivery_Time_Days=('Delivery_Time_Days', 'mean'),
    Delivered_Ratio=('Order_Status', lambda x: (x == 'Delivered').mean())
).reset_index()
display(delivery_by_segment)

delivery_by_channel = orders_summary.groupby('Sales_Channel').agg(
    Avg_Delivery_Time_Days=('Delivery_Time_Days', 'mean'),
    Delivered_Ratio=('Order_Status', lambda x: (x == 'Delivered').mean())
).reset_index()
display(delivery_by_channel)

```

```python
# 5. Answer Question 4: Customer repeat behavior and order value
# Calculate segment performance metrics
segment_metrics = orders_summary.groupby('Customer_Segment').agg(
    Total_Orders=('Order_ID', 'count'),
    Avg_Order_Value_EGP=('Total_Order_Sales_EGP', 'mean'),
    Total_Sales_EGP=('Total_Order_Sales_EGP', 'sum')
).sort_values(by='Avg_Order_Value_EGP', ascending=False).reset_index()

# Calculate overall repeat customer statistics
customer_stats = orders_summary.groupby('Customer_ID').agg(
    Total_Orders=('Order_ID', 'count'),
    Avg_Spend_EGP=('Total_Order_Sales_EGP', 'mean')
).reset_index()

total_customers = len(customer_stats)
repeat_customers_df = customer_stats[customer_stats['Total_Orders'] > 1]
repeat_count = len(repeat_customers_df)
repeat_percentage = (repeat_count / total_customers) * 100

print("=== Customer Segment Performance ===")
display(segment_metrics)

print("\n=== Repeat Purchase Behavior Summary ===")
print(f"Total Unique Customers: {total_customers}")
print(f"Repeat Customers (Ordered > 1 time): {repeat_count} ({repeat_percentage:.2f}%)")
print(f"Average Order Value (AOV) of Repeat Customers: {repeat_customers_df['Avg_Spend_EGP'].mean():.2f} EGP")
```

```python
# 6. Answer Question 5: Which customer or product action should the business test next?

# 1. Product Action Test: Identify high-sales but low-margin product or high-margin low-sales product
product_margins = order_items_clean.groupby('Product').agg(
    Total_Sales_EGP=('Line_Sales_EGP', 'sum'),
    Total_Profit_EGP=('Line_Profit_EGP', 'sum'),
    Total_Quantity=('Quantity', 'sum')
).reset_index()
product_margins['Profit_Margin'] = product_margins['Total_Profit_EGP'] / product_margins['Total_Sales_EGP']

print("=== Product Action Matrix (Sorted by Profit Margin) ===")
display(product_margins.sort_values(by='Profit_Margin', ascending=False))

# 2. Customer Action Test: Segment behavior and average order values
segment_behavior = orders_summary.groupby('Customer_Segment').agg(
    Avg_Order_Value=('Total_Order_Sales_EGP', 'mean'),
    Total_Orders=('Order_ID', 'count'),
    Avg_Shipping_Paid=('Shipping_Fee_EGP', 'mean')
).reset_index()

print("\n=== Customer Segment Metrics ===")
display(segment_behavior)
```
